*Q7. Sum of weights*

```python
X = df.loc[
    df["origin"] == "Asia",
    ["vehicle_weight", "model_year"]
    ].head(7).values
```

```python
XTX = X.T.dot(X)
```

```python
XTX_inv = np.linalg.inv(XTX)
```

```python
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
```

```python
w = XTX_inv.dot(X.T).dot(y)
```

```python
w
```

```python
w.sum()
```

*The sum of the elements of the result is 0.369*
